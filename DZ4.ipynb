```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df=pd.read_csv ('train.csv')
```

```python
#Первые пять строк датасета
df.head()
```

```python
#Распределение выживших
df_Survived=df['Survived'].value_counts()
y1= df_Survived[0]
y2= df_Survived[1]
x = ['Survived', 'Not survived']
y = [y1, y2]
fig, ax = plt.subplots()
ax.bar(x=x, height=y)
plt.xlabel('Outcome of the crash', color='blue')
plt.ylabel('Number of passengers', color='blue')
plt.title('Distribution of survivors')
```

```python
#Распределение пассажиров по классам
df_Pclass=df['Pclass'].value_counts()
y1= df_Pclass[1]
y2= df_Pclass[2]
y3= df_Pclass[3]
x= ['1 class', '2 class', '3 class']
y = [y1, y2, y3]
fig, ax = plt.subplots()
ax.bar(x=x, height=y)
plt.xlabel('Classes', color='blue')
plt.ylabel('Number of passengers', color='blue')
plt.title('Distribution by classes')
```

```python
#Рапределение пассажиров по возрасту
df_Age=df['Age'].value_counts()
y=df_Age.index
x = df_Age
y = y
sns.scatterplot(x=x, y=y, hue = y)
plt.xlabel('Number of passengers', color='purple')
plt.ylabel('Age of passengers', color='purple')
plt.title('Distribution by age')
plt.legend(loc='best')
```

```python
#Распределение пассажиров по полу
f=0
m=0
for i in df['Sex']:
  if i=='female':
    f+=1
for i in df['Sex']:
  if i=='male':
    m+=1
y1= f
y2= m
labels = ['Female', 'Male']
value = [y1, y2]
fig, ax = plt.subplots()
ax.pie(value, labels=labels, autopct= ' %1.1f%%')
ax.legend()
plt.title('Distribution by sex')
```

```python
#Распределение пассажиров по количеству родных на борту
df_Parch=df['Parch'].value_counts()
y1= df_Parch[0]
y2= df_Parch[1]
y3= df_Parch[2]
y4= df_Parch[3]
y5= df_Parch[4]
y6= df_Parch[5]
y7= df_Parch[6]
x= ['0', '1', '2','3','4','5','6']
y = [y1, y2, y3, y4, y5, y6, y7]
fig, ax = plt.subplots()
ax.bar(x=x, height=y)
plt.xlabel('Number of relatives', color='blue')
plt.ylabel('Number of passengers', color='blue')
plt.title('Distribution by parch')
```

```python
#График типа boxplot для столбца Age
plt.figure()
df.boxplot(['Age'],showmeans=True, meanline=True)
plt.title('Boxplot for age')
```

Интерпретация полученного графика: медиана проходит в районе 27 лет, средний возраст пассажиров составил 30 лет.Основные 50% возраста пассажиров лежат в пределах 20-38 лет. Высчитанный минимум возраста пассажира составил несколько месяцев, а высчитанный максимум - около 64 лет. Часть пассажиров в возрасте попали в категорию выбросов.

```python
# Вычисление количества пассажиров в каждом классе в процентах
y1= df_Pclass[1]
y2= df_Pclass[2]
y3= df_Pclass[3]
y=y1+y2+y3
с1= round(y1/y*100)
с2= round(y2/y*100)
с3= round(y3/y*100)
# Вычисление количества мужчин/женщин в каждом классе в процентах
Female_1=df.loc[(df['Sex'] == 'female') & (df['Pclass'] == 1)]
female1=Female_1.shape[0]
Male_1=df.loc[(df['Sex'] == 'male') & (df['Pclass'] == 1)]
male1=Male_1.shape[0]
Female_2=df.loc[(df['Sex'] == 'female') & (df['Pclass'] == 2)]
female2=Female_2.shape[0]
Male_2=df.loc[(df['Sex'] == 'male') & (df['Pclass'] == 2)]
male2=Male_2.shape[0]
Female_3=df.loc[(df['Sex'] == 'female') & (df['Pclass'] == 3)]
female3=Female_3.shape[0]
Male_3=df.loc[(df['Sex'] == 'male') & (df['Pclass'] == 3)]
male3=Male_3.shape[0]
all_fm=female1+male1+female2+male2+female3+male3
fm1=female1+male1
fm2=female2+male2
fm3=female3+male3
f1=round(female1/all_fm*100)
f2=round(female2/all_fm*100)
f3=round(female3/all_fm*100)
m1=int(male1/all_fm*100)
m2=round(male2/all_fm*100)
m3=round(male3/all_fm*100)
#sunburst plot (первый уровень - количество пассажиров в каждом из классов, второй - количество женщин/мужчин в этом классе)
data = dict(
    categories = ['All classes', '1 class', '2 class', '3 class',
                  'Female_1','Male_1','Female_2','Male_2','Female_3','Male_3'],
    parent = ['', 'All classes', 'All classes', 'All classes', '1 class', '1 class', '2 class', '2 class', '3 class', '3 class',],
    value = [100, с1, с2, с3, f1, m1, f2, m2, f3, m3]
)
fig = px.sunburst(data, names='categories', parents='parent',values = 'value', branchvalues='total')
fig.show()
```
