```python
class Chocolate:
   def __init__(self,price:float,additive:str=''):
    self.price=price
    self.additive=additive
    if price<0:
      raise ValueError('Отрицательное значение цены!')
   def __str__(self):
      return f'Шоколадный батончик, {self.price} рублей'
   def show_my_chocolate (self):
    if self.additive!= '':
       print ('Шоколадный батончик, добавка:',self.additive)
    else:
      print('Обычный шоколадный батончик')
```

```python
Twix = Chocolate (-15.5, 'c карамелью')
Twix.show_my_chocolate()
print (Twix)
```

```python
Milka = Chocolate (90, 'c миндалем')
Milka.show_my_chocolate()
print (Milka)
```

```python
Milkyway = Chocolate (30)
Milkyway.show_my_chocolate()
print (Milkyway)
```
