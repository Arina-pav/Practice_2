```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv ('train.csv')
#информация о типах данных
df.info()
```

```python
#Первые пять строк датасета
df.head()
```

```python
#Расчет количества пропусков
df.isna().sum()
```

```python
#Среднее значение возраста
mean_Age=df['Age'].mean()
print(mean_Age)
```

```python
#Среднее значение стоимости
mean_Fare=df['Fare'].mean()
print(mean_Fare)
```

```python
#Количество выживших и погибших
df['Survived'].value_counts()
```

```python
#Расчет количества пассажиров в каждом классе и число выживших в них в виде двух датасетов
df_Pclass=df.groupby("Pclass")["Survived"].sum()
all_Pclass=df["Pclass"].value_counts()
print (df_Pclass)
print (all_Pclass)
```

```python
#Объединение датасетов и его сортировка по индексу
two_df = pd.concat([all_Pclass, df_Pclass],axis=1).sort_index()
print(two_df)
```

```python
#Процент выживаемости у каждого класса пассажиров
two_df['Survived']/two_df['count']*100
```

```python
#Датасет женщин
Women=df.loc[df["Sex"] == 'female', "Name"]
```

```python
#Датасет мужчин
Men=df.loc[df["Sex"] == 'male', "Name"]
```

```python
#Самое популярное женское имя на корабле
Women_name=Women.str.split(' ',expand=True,n=3)[2].value_counts()
print(Women_name.index[0]+' '+Women_name.astype('str').iloc[0]+' раз встречается среди женщин')
```

```python
#Cамое популярное мужское имя на корабле
Men_name=Men.str.split(' ',expand=True,n=3)[2].value_counts()
print(Men_name.index[0]+' '+Men_name.astype('str').iloc[0]+' раз встречается среди мужчин')
```

```python
#Cамое популярное женское имя в первом классе
Women_1=df.loc[(df["Sex"] == 'female') & (df["Pclass"] == 1), "Name"]
Women_name_1=Women_1.str.split(' ',expand=True,n=3)[2].value_counts()
print(Women_name_1.index[0]+' '+Women_name_1.astype('str').iloc[0]+' раз встречается среди женщин первого класса')
```

```python
#Cамое популярное мужское имя в первом классе
Men_1=df.loc[(df["Sex"] == 'male') & (df["Pclass"] == 1), "Name"]
Men_name_1=Men_1.str.split(' ',expand=True,n=3)[2].value_counts()
print(Men_name_1.index[0]+' '+Men_name_1.astype('str').iloc[0]+' раз встречается среди мужчин первого класса')
```
